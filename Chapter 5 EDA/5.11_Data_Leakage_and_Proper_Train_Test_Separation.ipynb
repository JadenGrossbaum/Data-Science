### 5.11.1 Leakage definition

Data leakage occurs when training or evaluation uses information that would not be available for the intended prediction. Types include target leakage, preprocessing leakage, temporal leakage, group leakage, and feature-selection leakage.

### 5.11.2 Point-in-time correctness

For each feature, check event time **and** availability time. A claim recorded after the prediction timestamp is unavailable even if its event date is earlier.

### 5.11.3 Train-only preprocessing

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import make_classification
X,y=make_classification(n_samples=600,n_features=8,random_state=42)
Xtr,Xte,ytr,yte=train_test_split(X,y,stratify=y,random_state=42)
model=make_pipeline(SimpleImputer(),StandardScaler(),LogisticRegression(max_iter=1000))
model.fit(Xtr,ytr)
print("test accuracy",model.score(Xte,yte))
```

### 5.11.4 Evaluation design

Use group-aware splitting for related observations, temporal splitting for forecasting, and label-maturity gaps for overlapping prediction windows. Keep the final test set untouched.
