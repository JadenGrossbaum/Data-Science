### 5.14.1 From notebook to production

A production ML system needs data contracts, validation, versioned artifacts, prediction logging, label monitoring, drift detection, and defined operational responses.

### 5.14.2 Drift taxonomy

**Covariate shift:** $P_{train}(X)\ne P_{prod}(X)$ while $P(Y|X)$ remains stable. **Concept drift:** $P_{train}(Y|X)\ne P_{prod}(Y|X)$. **Prediction drift:** distribution of model outputs changes.

### 5.14.3 KS and PSI

KS: $D_{n,m}=\sup_x|F_n(x)-G_m(x)|$. PSI for reference bin proportions $p_i$ and current proportions $q_i$: $\sum_i(q_i-p_i)\log(q_i/p_i)$. Binning and thresholds require contextual calibration.

### 5.14.4 Python drift example

```python
import numpy as np
from scipy.stats import ks_2samp
rng=np.random.default_rng(42)
reference=rng.normal(0,1,2000)
production=rng.normal(.5,1,2000)
result=ks_2samp(reference,production)
print("KS statistic",result.statistic,"p-value",result.pvalue)
```

### 5.14.5 Monitor missingness

```python
import pandas as pd
ref=pd.DataFrame({"premium":[1000,1200,1300,None]})
prod=pd.DataFrame({"premium":[None,None,1400,None]})
print(pd.DataFrame({"reference":ref.isna().mean(),"production":prod.isna().mean()}))
```
