### 5.2.1 Observation units and schemas

A row might represent a customer, policy, transaction, or policy-month. Multiple rows per entity change how you split data and interpret duplicates.

### 5.2.2 Feature types

**Numerical:** continuous (premium) or discrete (claims count). **Categorical:** nominal (region) or ordinal (risk tier). **Identifiers:** policy ID is often numeric but not a quantitative predictor. **Temporal:** timestamps, durations, and lookback windows.

### 5.2.3 Python inspection

```python
import pandas as pd
df = pd.DataFrame({"policy_id":[101,102,103], "age":[29,45,53], "policy_type":["Auto","Home","Life"], "cancelled":[0,1,0]})
print(df.dtypes)
print(df.nunique())
print(df.select_dtypes(include="number").columns.tolist())
```

### 5.2.4 Schema caveats

A numeric ID must not be treated as continuous merely because it has an integer dtype. Validate units, allowable ranges, unique keys, and missingness.
