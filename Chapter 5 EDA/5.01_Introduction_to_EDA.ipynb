### 5.1.1 What EDA accomplishes

Exploratory data analysis (EDA) investigates the structure, quality, distributions, and relationships of a dataset before and during model development. It is not merely plotting charts: its purpose is to make defensible modeling decisions.

**Workflow:** define the prediction problem → inspect observations and schema → audit data quality → investigate distributions and relationships → protect the holdout set → document findings.
