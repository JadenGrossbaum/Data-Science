### 5.8.1 Outliers and anomalies

The 1.5-IQR fences are $Q_1-1.5IQR$ and $Q_3+1.5IQR$. Z-scores flag observations far from the mean but are sensitive to extreme values and distribution shape.

### 5.8.2 Outlier versus error

An unusually high insurance premium may be valid; a negative premium may violate business rules. Investigate rather than automatically delete.

### 5.8.3 Python example

```python
import pandas as pd
x=pd.Series([10,12,13,15,16,18,20,22,100])
q1,q3=x.quantile([.25,.75]);iqr=q3-q1
low,high=q1-1.5*iqr,q3+1.5*iqr
print("fences",low,high)
print("flagged",x[(x<low)|(x>high)].tolist())
```

### 5.8.4 Treatment options

Verify source values, apply domain rules, consider robust models or transformations, and evaluate any winsorization within the training process.
