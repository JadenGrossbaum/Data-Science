### 5.7.1 Missingness mechanisms

**MCAR:** missingness independent of observed and unobserved values. **MAR:** conditional on observed information, missingness does not depend on the missing value. **MNAR:** missingness may depend on unobserved values even after conditioning. These are assumptions about the data-generating process, not categories that can always be identified from a plot.

### 5.7.2 Missingness rates

For feature $j$: $M_j=\frac1n\sum_i\mathbf{1}(X_{ij}\text{ missing})$. Compare missingness across groups and time; consider whether missingness is informative.

### 5.7.3 Python example

```python
import numpy as np,pandas as pd
df=pd.DataFrame({"income":[50000,np.nan,70000,np.nan],"age":[25,38,52,43],"cancelled":[0,1,0,1]})
print(pd.DataFrame({"missing_n":df.isna().sum(),"missing_pct":100*df.isna().mean()}))
print(df.assign(income_missing=df.income.isna()).groupby("income_missing")["cancelled"].agg(["mean","count"]))
```

### 5.7.4 Safe imputation

Fit median/mode or other learned imputers only on training data, preferably inside a pipeline. Missingness indicators may help. Do not assume deletion is harmless.
