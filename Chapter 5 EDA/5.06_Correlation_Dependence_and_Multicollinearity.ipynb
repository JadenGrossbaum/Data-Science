### 5.6.1 Correlation

Pearson correlation is $\rho_{XY}=\frac{\operatorname{Cov}(X,Y)}{\sigma_X\sigma_Y}$. Zero Pearson correlation does not imply independence. Spearman correlation measures rank-based monotonic association.

### 5.6.2 Multicollinearity

Highly correlated predictors can destabilize individual linear-model coefficients. Variance inflation factor: $VIF_j=1/(1-R_j^2)$, where $R_j^2$ is obtained by regressing predictor $j$ on the other predictors.

### 5.6.3 Python example

```python
import numpy as np,pandas as pd,seaborn as sns,matplotlib.pyplot as plt
rng=np.random.default_rng(42); x=rng.normal(size=300)
df=pd.DataFrame({"income":x,"income_proxy":2*x+rng.normal(0,.2,300),"claims":rng.poisson(2,300)})
print(df.corr(numeric_only=True))
sns.heatmap(df.corr(numeric_only=True),annot=True,center=0,cmap="coolwarm");plt.show()
```
