### 5.4.1 Univariate numerical analysis

Use histograms, KDE curves, ECDFs, boxplots, and quantile tables to study a single numerical feature. Choose bin widths deliberately; KDE smoothing can hide discrete structure.

### 5.4.2 Univariate categorical analysis

Use counts and proportions for categories. Examine cardinality, rare levels, and inconsistent capitalization. Missingness can be a category worth reporting.

### 5.4.3 Visual examples

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
rng=np.random.default_rng(42)
df=pd.DataFrame({"premium":rng.lognormal(7.4,.5,500),"policy_type":rng.choice(["Auto","Home","Life"],500)})
fig,axes=plt.subplots(1,2,figsize=(11,4))
sns.histplot(df,x="premium",kde=True,ax=axes[0])
sns.countplot(df,x="policy_type",ax=axes[1])
plt.tight_layout(); plt.show()
```

### 5.4.4 Reading plots

Ask whether tails, multiple modes, spikes, missingness, and implausible values suggest distinct subpopulations or data errors.
