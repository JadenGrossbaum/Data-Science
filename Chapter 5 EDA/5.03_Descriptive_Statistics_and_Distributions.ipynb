### 5.3.1 Central tendency and dispersion

Sample mean: $\bar x=\frac1n\sum_i x_i$. Sample variance: $s^2=\frac1{n-1}\sum_i(x_i-\bar x)^2$. Standard deviation: $s=\sqrt{s^2}$. The median is robust to extreme observations.

### 5.3.2 Quantiles and skewness

The interquartile range is $IQR=Q_3-Q_1$. A long right tail indicates positive skew; means can exceed medians. Log transformations may help some models but should be evaluated rather than applied automatically.

### 5.3.3 Python calculations

```python
import pandas as pd
x = pd.Series([4,6,8,10,12,100])
print(x.describe())
print("mean",x.mean(),"median",x.median(),"variance",x.var(ddof=1))
print("skew",x.skew(),"IQR",x.quantile(.75)-x.quantile(.25))
```

### 5.3.4 Interpretation

Always inspect units, sample sizes, missingness, and extreme observations before interpreting a summary statistic.
