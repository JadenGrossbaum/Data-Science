### 5.10.1 Class imbalance

When positives are rare, majority-class accuracy is misleading. Positive prevalence is $\hat\pi=\frac1n\sum_i y_i$. Evaluate confusion matrices, precision, recall, F1, ROC-AUC, and average precision.

### 5.10.2 Metric formulas

Precision $=TP/(TP+FP)$; recall $=TP/(TP+FN)$; F1 $=2PR/(P+R)$. Threshold selection should reflect the business objective.

### 5.10.3 Python example

```python
import numpy as np
from sklearn.metrics import classification_report,average_precision_score,roc_auc_score
y=np.array([0]*90+[1]*10)
scores=np.r_[np.linspace(.01,.4,90),np.linspace(.2,.9,10)]
print(classification_report(y,(scores>=.5).astype(int),zero_division=0))
print("AP",average_precision_score(y,scores),"ROC-AUC",roc_auc_score(y,scores))
```
