### 5.9.2 Quality checks

```python
import pandas as pd,numpy as np
df=pd.DataFrame({"policy_id":[1,2,2],"age":[32,250,45],"premium":[1000,-20,np.nan],"type":["Auto","Home","home"]})
print("duplicate ids",df.policy_id.duplicated().sum())
print("invalid age",(~df.age.between(18,100)).sum())
print("invalid premiums",(df.premium.notna()&(df.premium<=0)).sum())
print("missing",df.isna().sum())
print(df.type.str.strip().str.title().value_counts())
```
