### 5.5.1 Bivariate analysis

Analyze numeric–numeric relationships with scatterplots; categorical–numeric relationships with boxplots or group summaries; categorical–categorical relationships with contingency tables and conditional proportions.

### 5.5.2 Multivariate structure

Relationships can reverse or disappear after conditioning on other features. Consider confounding, interactions, nonlinear patterns, and subgroup behavior.

### 5.5.3 Python example

```python
import numpy as np,pandas as pd,seaborn as sns,matplotlib.pyplot as plt
rng=np.random.default_rng(42)
df=pd.DataFrame({"tenure":rng.uniform(0,10,400),"policy_type":rng.choice(["Auto","Home"],400)})
df["premium"]=1200+100*df["tenure"]+rng.normal(0,300,400)
df["cancelled"]=(rng.random(400)<.15).astype(int)
sns.scatterplot(df,x="tenure",y="premium",hue="policy_type",alpha=.6);plt.show()
print(df.groupby("policy_type")["cancelled"].agg(["mean","count"]))
```

### 5.5.4 Interpretation

Association is not causation. Small subgroup counts make apparent differences unstable.
