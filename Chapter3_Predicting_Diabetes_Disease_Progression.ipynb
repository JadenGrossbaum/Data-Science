### Part 3 — Training Our First Machine Learning Model
Now we can begin estimating the unknown function $f(X)$.

#### 3.1 — Choosing our model
We'll start with multiple linear regression because we have 10 predictors and one continuous target.
Our model assumes:
$[\hat Y=\hat\beta_0+\hat\beta_1X_1+\cdots+\hat\beta_{10}X_{10}\]$

Where:
- $\hat\beta_0$ is the intercept.
- $\hat\beta_j$ is the estimated coefficient for predictor \(j\).
- $X_j$ is the corresponding feature value.
The algorithm estimates the coefficients by minimizing the sum of squared residuals:
$[\boxed{\min_{\beta_0,\ldots,\beta_{10}}\sum_{i=1}^{n}\left(y_i-\beta_0-\sum_{j=1}^{10}\beta_jx_{ij}\right)^2}\]$

This is called ordinary least squares (OLS).
A residual is the difference between an observed value and its fitted prediction:
$[e_i=y_i-\hat y_i\]$

The smaller the squared residuals, the better the model fits the training observations according to this objective.

### Part 5 — Evaluating Regression Performance
We'll introduce three metrics:
1. Mean Squared Error (MSE)
2. Root Mean Squared Error (RMSE)
3. Coefficient of Determination (\(R^2\))
For now, we'll calculate these using the training data. Later, we'll compare models using cross-validation and evaluate our final model on the held-out test set.

#### 5.1 — Mean Squared Error (MSE)
Mean Squared Error measures the average squared difference between actual and predicted values.
$[\boxed{MSE=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat y_i)^2}]$

How it works
For every observation:
1. Calculate the residual: $y_i-\hat y_i$.
2. Square the residual.
3. Add the squared residuals.
4. Divide by the number of observations.

Why square the residuals?
- Prevents positive and negative residuals from canceling.
- Penalizes larger prediction errors more heavily.
- Produces a differentiable objective that is convenient for optimization.

Example
Suppose three patients have the following outcomes:
Actual	Predicted	Residual	Squared residual
100	    90	        10	        100
150	    170	        -20	        400
200	    195	        5	        25

Then:
$[MSE=\frac{100+400+25}{3}=\boxed{175}]$

A lower MSE indicates better predictive accuracy on the evaluated observations.
However, MSE is measured in squared units of the target, making it less intuitive to interpret directly.

#### 5.2 — Root Mean Squared Error (RMSE)
RMSE is simply the square root of MSE.
$[\boxed{RMSE=\sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat y_i)^2}}]$

Using our previous example:
$[RMSE=\sqrt{175}\approx\boxed{13.23}]$

RMSE is expressed in the same units as the target variable.
This makes it easier to interpret the typical magnitude of prediction errors, although RMSE is not exactly the average absolute error.
For example, an RMSE of 50 means the model's prediction errors have a root-mean-square magnitude of 50 target units.

#### 5.3 — Coefficient of Determination — $R^2$
Unlike MSE and RMSE, which measure prediction error, $R^2$ measures performance relative to a baseline model that always predicts the mean of the observed targets.
$[\boxed{R^2=1-\frac{SS_{\text{res}}}{SS_{\text{tot}}}}]$

Where:
$[SS_{\text{res}}=\sum_{i=1}^{n}(y_i-\hat y_i)^2]$

$[SS_{\text{tot}}=\sum_{i=1}^{n}(y_i-\bar y)^2]$

Here, $\bar y$ is the mean target value in the dataset being evaluated.

Interpreting $R^2$
$R^2$	    Interpretation
1.0	        Perfect predictions
0.8	        80% reduction in squared error relative to the mean baseline
0.5	        50% reduction in squared error relative to the mean baseline
0.0	        Same squared error as predicting the mean
Negative	Worse than predicting the evaluation sample's mean

In ordinary least squares with an intercept, training $R^2$ is generally between 0 and 1. On unseen data, $R^2$ can be negative.
One important point: A high $R^2$ does not automatically mean a model generalizes well. We still need validation and test performance.

### Part 6 — Cross-Validation
We're going to introduce one of the most important tools in applied machine learning: K-fold cross-validation.

#### 6.1 — Why use cross-validation?
We currently have:
- 353 training observations.
- 89 test observations.
We want to estimate how well our model performs on unseen data without using our final test set for model selection.
One option would be to divide the training data into a smaller training set and a validation set.
But with only 353 training observations, we'd like to use our limited data efficiently.
K-fold cross-validation helps us do this.

#### 6.2 — How K-fold cross-validation works
Suppose we choose:
$K=5$

We divide our 353 training observations into five approximately equal folds.
5-fold cross-validation                 Each row = one iteration
Fold 1       Val       Train       Train       Train       Train
Fold 2       Train     Val         Train       Train       Train
Fold 3       Train     Train       Val         Train       Train
Fold 4       Train     Train       Train       Val         Train
Fold 5       Train     Train       Train       Train       Val
                Training fold       Validation fold

During each iteration:
1. Train a fresh model on four folds.
2. Predict the outcomes of the remaining fold.
3. Calculate the validation error.
4. Repeat until every fold has been used for validation.

The final cross-validation score is the average of the five validation scores.
$[CV_{(5)}=\frac{1}{5}\sum_{k=1}^{5}MSE_k]$

This provides an estimate of how well the learning procedure performs on observations not used to fit each fold's model.

#### 6.3 — Implementing cross-validation in scikit-learn
We can use:
`from sklearn.model_selection import cross_val_score`

The function follows this structure:
`cross_val_score(    
    estimator,    
    X,    
    y,    
    cv=5,    
    scoring="neg_mean_squared_error"
)`

Let's understand the arguments.
Argument	    Meaning
`estimator`	    Model to evaluate
`X`	            Features used for cross-validation
`y`	            Corresponding targets
`cv=5`	        Five-fold cross-validation
`scoring`	        Metric used to evaluate predictions

Why is the scoring metric negative?
Scikit-learn uses a convention where higher scores are better.
But MSE is an error metric, where lower values are better.
Therefore:
`scoring="neg_mean_squared_error"`
returns the negative MSE for each fold.
To recover the ordinary positive MSE, multiply the scores by $-1$.
Also, `cross_val_score()` fits fresh copies of the estimator internally. It does not reuse the coefficients previously learned by `model.fit(X_train, y_train)`.

### Part 7 — Investigating Model Flexibility
Now we're going to investigate one of Chapter 3's central concepts: the bias–variance tradeoff.
So far, we've used ordinary linear regression.
Our model assumes:
$[\hat Y=\hat\beta_0+\sum_{j=1}^{10}\hat\beta_jX_j]$

This assumes an additive, linear relationship between the predictors and disease progression.
But what if the true relationship is nonlinear?
For example, the relationship between BMI and disease progression might involve curvature that a straight-line term cannot represent.

#### 7.1 — Polynomial regression
We can increase model flexibility by introducing polynomial features.
For a single predictor:
Linear model:
$[\hat Y=\beta_0+\beta_1X]$

Quadratic model:
$[\hat Y=\beta_0+\beta_1X+\beta_2X^2]$

Cubic model:
$[\hat Y=\beta_0+\beta_1X+\beta_2X^2+\beta_3X^3]$

Even though these models contain nonlinear terms in $X$, they are still linear in their coefficients, so ordinary linear regression can estimate them.

#### 7.2 — The tradeoff
Increasing polynomial degree gives the model more flexibility.
This can reduce bias by allowing the model to capture nonlinear patterns.
However, increasing flexibility also creates opportunities to fit noise in the training data.
Our objective is to determine whether the additional flexibility actually improves cross-validation performance.

#### 7.3 — Introducing `PolynomialFeatures`
Scikit-learn provides:
`from sklearn.preprocessing import PolynomialFeatures`

This transformer creates polynomial and interaction features.
For example, given two features $X_1$ and $X_2$, a degree-2 transformation with the default settings creates:
$[X_1,\quad X_2,\quad X_1^2,\quad X_1X_2,\quad X_2^2]$

plus a constant feature.
With 10 original features, degree 2 produces 66 output columns when include_bias=True, or 65 when include_bias=False.
That's a substantial increase from our original 10 predictors.

#### 7.4 — Why we use a pipeline
We need to ensure that feature transformations are performed consistently during cross-validation.
Scikit-learn provides:
`from sklearn.pipeline import Pipeline`

A pipeline combines preprocessing and modeling into one estimator.
For example:
`pipeline = Pipeline([    
    ("poly", PolynomialFeatures(degree=2, include_bias=False)),    
    ("linear", LinearRegression())])
`

The pipeline first generates polynomial features and then fits linear regression on the transformed data.
During cross-validation, scikit-learn fits a fresh copy of the complete pipeline on each training fold.
For this particular deterministic polynomial transformation, there are no data-dependent statistics to learn. But the pipeline structure becomes especially important later when we introduce scaling, imputation, and other preprocessing operations.

### Part 8 — Introducing Regularization
Now we're going to investigate whether we can improve polynomial regression by controlling model complexity.

#### 8.1 — Why regularization?
Ordinary least squares minimizes:
$[\boxed{RSS=\sum_{i=1}^{n}(y_i-\hat y_i)^2}]$

It attempts to find coefficients that minimize squared training errors.
But when we introduce many polynomial features, the model has more opportunities to fit sample-specific noise.
Regularization adds a penalty to the optimization objective.
For example, Ridge regression minimizes:
$[\boxed{\sum_{i=1}^{n}(y_i-\hat y_i)^2+\alpha\sum_{j=1}^{p}\beta_j^2}]$

Here:
- $\alpha$ controls the strength of regularization.
- $\beta_j$ represents a feature coefficient.
- The intercept is not penalized.
As $\alpha$ increases, the penalty for large coefficients increases.
This generally reduces model flexibility and variance, potentially at the cost of increasing bias.

#### 8.2 — Ridge versus ordinary linear regression
Property	                        Linear regression	                    Ridge regression
Minimizes squared error	            Yes	                                    Yes
Penalizes coefficients	            No	                                    Yes
Controls coefficient magnitude	    No explicit penalty	                    Yes
Can reduce overfitting	            Limited	                                Potentially
Introduces bias	                    Not through an explicit penalty	        Yes
An important distinction: Ridge does not remove polynomial features. Instead, it discourages the model from relying on excessively large coefficients.

#### Part 9 — Hyperparameter Tuning

#### 9.1 — Parameters versus hyperparameters
This distinction is fundamental to machine learning.
Model parameters are learned from training data.
For Ridge regression, these include:
$\beta_0,\beta_1,\ldots,\beta_p$

Hyperparameters are configuration choices that control the learning process.
For our Ridge model, one hyperparameter is:
$\alpha$

We choose $\alpha$ before fitting the model.
The learning algorithm then estimates the coefficients subject to the chosen regularization strength.

Why can RMSE remain almost identical while \(R^2\) decreases?
Because $R^2$ depends on both prediction error and the total variation in the target values:
$[R^2=1-\frac{\sum_i(y_i-\hat y_i)^2}{\sum_i(y_i-\bar y)^2}]$

The training and test sets can have different amounts of target variation.
Therefore, similar RMSE values do not guarantee similar $R^2$ values.
